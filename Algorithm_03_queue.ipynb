```python
from collections import deque
queue = deque()

print('시작')
    
while True:
    #메뉴 프린트
    #메뉴 번호 입력
    menu = int(input('>>'))
    
    if menu == 0: break
    elif menu == 1: #enqueue
        data = int(input('데이터 :'))
        queue.append(data)
    elif menu == 2: # dequeue
        if queue:
            data = queue.popleft()
            print('가져온 데이터:', data)
        else:
            print('Queue가 비어 있습니다')
    elif menu == 2: # peek
        if queue:
            print(queue[0])
        else:
            print('Queue가 비어 있습니다')
    
    print(queue)
    
    
print('종료')             
```
